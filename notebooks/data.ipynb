```python
import torchvision.transforms as T

CIFAR10_MEAN = (0.4914, 0.4822, 0.4465)
CIFAR10_STD  = (0.2470, 0.2435, 0.2616)

train_transform = T.Compose([
    T.RandomCrop(32, padding=4, padding_mode='reflect'),
    T.RandomHorizontalFlip(p=0.5),
    T.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    T.ToTensor(),
    T.Normalize(CIFAR10_MEAN, CIFAR10_STD),
    T.RandomErasing(p=0.25, scale=(0.02, 0.15)),  # applied after normalize, on tensor
])

test_transform = T.Compose([
    T.ToTensor(),
    T.Normalize(CIFAR10_MEAN, CIFAR10_STD),
])
```

```python
from torchvision.datasets import CIFAR10
from torch.utils.data import DataLoader

train_set = CIFAR10(root='.', train=True, download=False, transform=train_transform)
test_set  = CIFAR10(root='.', train=False, download=False, transform=test_transform)

print(len(train_set), len(test_set))
```
